# 03 — Core Principles of MLOps

This notebook illustrates the major MLOps principles with practical Python examples.

## Learning Objectives

- Reproducibility
- Version Control concepts
- Experiment Tracking
- Automation
- CI/CD concepts
- Monitoring
- Governance

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)
```

## 1. Reproducibility

```
seed = 42
np.random.seed(seed)

sample = np.random.normal(size=1000)

print('Mean:', sample.mean())
print('Std :', sample.std())
```

### Exercise

Change the seed and compare the results.

## 2. Versioning Example

```
dataset_v1 = pd.DataFrame({
    'income':[30000,50000,70000],
    'target':[0,0,1]
})

dataset_v2 = pd.DataFrame({
    'income':[30000,50000,70000,90000],
    'target':[0,0,1,1]
})

print('Rows v1:', len(dataset_v1))
print('Rows v2:', len(dataset_v2))
```


Discussion:

- Which version trained the model?
- Can results be reproduced without dataset versioning?


## 3. Experiment Tracking

```
experiments = pd.DataFrame({
    'run_id':['run_1','run_2','run_3'],
    'learning_rate':[0.01,0.05,0.1],
    'auc':[0.78,0.82,0.80]
})

experiments.sort_values('auc', ascending=False)
```


Typical information stored by MLflow:

- Parameters
- Metrics
- Artifacts
- Dataset version
- User
- Timestamp


## 4. Automation

```
pipeline_steps = [
    'Data ingestion',
    'Validation',
    'Feature engineering',
    'Training',
    'Evaluation',
    'Deployment'
]

for step in pipeline_steps:
    print(step)
```


Automation reduces:

- Manual errors
- Deployment delays
- Reproducibility issues


## 5. Continuous Integration

```
tests = {
    'unit_tests': 'PASS',
    'schema_validation': 'PASS',
    'code_quality': 'PASS'
}

pd.DataFrame(
    list(tests.items()),
    columns=['check','status']
)
```


Typical CI actions:

- Run tests
- Validate data schema
- Build artifacts
- Prepare deployment package


## 6. Continuous Delivery

```
environments = [
    'Development',
    'Testing',
    'Staging',
    'Production'
]

pd.DataFrame({'environment': environments})
```


Promotion flow:

Development → Testing → Staging → Production


## 7. Monitoring Example

```
months = np.arange(1,13)

auc = [0.84,0.84,0.83,0.83,0.82,0.81,0.79,0.77,0.74,0.72,0.69,0.66]

plt.figure(figsize=(8,4))
plt.plot(months, auc, marker='o')
plt.title('Model Performance Degradation')
plt.xlabel('Month')
plt.ylabel('AUC')
plt.grid(True)
plt.show()
```


Questions:

- When should an alert be triggered?
- When should retraining occur?


## 8. Governance

```
governance = pd.DataFrame({
    'artifact':[
        'Dataset',
        'Model',
        'Validation Report',
        'Deployment Record'
    ],
    'required':['Yes','Yes','Yes','Yes']
})

governance
```


Governance ensures:

- Auditability
- Traceability
- Compliance
- Accountability
